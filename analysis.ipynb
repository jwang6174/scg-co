```python
import ast
import json
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
import numpy as np
import os
import pandas as pd
import scipy.stats as stats
from pathlib import Path

# pd.set_option('display.max_rows', None)
# pd.set_option('display.max_columns', None)


def flatten_maclab_meas_dict(df):
    expanded = pd.json_normalize(df['maclabMeas'])
    df = pd.concat([df.drop(columns=['maclabMeas']), expanded], axis=1)
    return df

def get_full_df():
    rows = []
    filenames = set()
    dir_path = os.path.join('/', 'Users', 'jessewang', 'SCG-RHC', 'data', 'processed_data')
    for filename in os.listdir(dir_path):
        if filename.endswith('.dat') or filename.endswith('.hea'):
            filenames.add(Path(filename).stem)

    for filename in filenames:
        filepath = os.path.join(dir_path, f'{filename}.json')
        with open(filepath, 'r') as f:
            data = json.load(f)
            data['name'] = filename
            rows.append(data)

    df = pd.DataFrame(rows)
    return df

def compare_num_col(df1, df2, col):
    _, p_val = stats.ttest_ind(df1[col], df2[col], equal_var=False)
    print(col, p_val)

def compare_num(df1, df2):
    compare_num_col(df1, df2, 'age')
    compare_num_col(df1, df2, 'bmi')
    compare_num_col(df1, df2, 'bsa')
    compare_num_col(df1, df2, 'RAM')
    compare_num_col(df1, df2, 'PAM')
    compare_num_col(df1, df2, 'PCWM')
    compare_num_col(df1, df2, 'PCWHR')
    compare_num_col(df1, df2, 'SVmL/beat')
    compare_num_col(df1, df2, 'Avg. COmL/min')

def flatten_maclab_meas(df):
    df.loc[:, 'maclabMeas_dict'] = df['maclabMeas'].apply(ast.literal_eval)
    expanded = pd.json_normalize(df['maclabMeas_dict'])
    expanded = expanded.reset_index(drop=True)
    df = pd.concat([df.drop(columns=['maclabMeas', 'maclabMeas_dict']), expanded], axis=1)
    return df
    
def add_bmi(df):
    bmi = df['weight'] / ((df['height']/100) ** 2)
    df.loc[:, 'bmi'] = bmi
    return df

def add_bsa(df):
    bsa = np.sqrt(df['height'] * df['weight'] / 3600)
    df.loc[:, 'bsa'] = bsa
    return df

def add_ci(df):
    real_ci = df['real']/df['bsa']
    pred_ci = df['pred']/df['bsa']
    df.loc[:, 'real_ci'] = real_ci
    df.loc[:, 'pred_ci'] = pred_ci
    return df

def add_color(co):
    if co > 6.0:
        return 'indianred'
    else:
        return 'steelblue'

def preprocess(df):
    df['real'] /= 1000
    df['pred'] /= 1000
    df = flatten_maclab_meas(df)
    df = add_bmi(df)
    df = add_bsa(df)
    df = add_ci(df)
    df['color'] = df['real'].apply(add_color)
    df.columns = df.columns.str.strip()
    return df

def summarize_num_col(df, col):
    avg = df[col].mean()
    std = df[col].std()
    min_ = df[col].min()
    max_ = df[col].max()
    print(f'{col} avg', avg)
    print(f'{col} std', std)
    print(f'{col} min', min_)
    print(f'{col} max', max_)
    print('-' * 30)

def summarize_ord_col(df, col):
    summary = pd.Series(df[col]).value_counts().sort_index()
    percent = summary / summary.sum() * 100
    print(pd.DataFrame({'Count': summary, 'Percent': percent.round(1)}))
    print('-' * 30)

def summarize(df):
    if 'real' in df.columns:
        summarize_num_col(df, 'real')
    if 'pred' in df.columns:
        summarize_num_col(df, 'pred')
    add_bmi(df)
    add_bsa(df)
    summarize_num_col(df, 'age')
    summarize_ord_col(df, 'gender')
    summarize_num_col(df, 'bmi')
    summarize_num_col(df, 'bsa')
    summarize_ord_col(df, 'NYHAC')
    summarize_num_col(df, 'RAM')
    summarize_num_col(df, 'PAM')
    summarize_num_col(df, 'PCWM')
    summarize_num_col(df, 'PCWHR')
    summarize_num_col(df, 'SVmL/beat')
    summarize_num_col(df, 'Avg. COmL/min')
    if 'real_ci' in df.columns:
        summarize_num_col(df, 'real_ci')
    summarize_ord_col(df, 'outpatient')
    
def get_rmse(df, col_x, col_y):
    rmse_samples = []
    n_iterations = 10_000
    n = len(df)
    for _ in range(n_iterations):
        sample = df.sample(n, replace=True)
        rmse_val = np.sqrt(np.mean((sample[col_x] - sample[col_y])**2))
        rmse_samples.append(rmse_val)
    lower, upper = np.percentile(rmse_samples, [2.5, 97.5])
    point = np.sqrt(np.mean((df[col_x] - df[col_y])**2))
    return point, lower, upper

def get_pearson(df, col_x, col_y, confidence=0.95):
    x = df[col_x]
    y = df[col_y]
    r, p = stats.pearsonr(x, y)
    n = len(x)
    if n <= 3:
        raise ValueError("At least 4 samples are needed for a CI.")
    z = np.arctanh(r) 
    se = 1 / np.sqrt(n - 3)
    z_crit = stats.norm.ppf(1 - (1 - confidence) / 2)
    z_ci_lower = z - z_crit * se
    z_ci_upper = z + z_crit * se
    return r, p, np.tanh(z_ci_lower), np.tanh(z_ci_upper)

def show_stats(df, y_true, y_pred):
    rmse, rmse_lower, rmse_upper = get_rmse(df, y_true, y_pred)
    rmse_percent = rmse / df[y_true].mean()
    pcc, pcc_p, pcc_lower, pcc_upper = get_pearson(df, y_true, y_pred)
    mean_values = (df[y_pred] + df[y_true]) / 2
    diff_values = df[y_pred] - df[y_true]
    mean_diff = np.mean(diff_values)
    std_diff = np.std(diff_values, ddof=1)
    loa_upper = mean_diff + 1.96 * std_diff
    loa_lower = mean_diff - 1.96 * std_diff
    print(f'Count = {len(df)}')
    print(f'RMSE = {rmse} (95% CInt: {rmse_lower}-{rmse_upper})')
    print(f'RMSE% = {rmse_percent}')
    print(f'PCC = {pcc} (95% CInt: {pcc_lower}-{pcc_upper}, p = {pcc_p})')
    print(f'Mean Diff = {mean_diff}')
    print(f'LoA Upper = {loa_upper}')
    print(f'LoA Lower = {loa_lower}')
    print()

def get_bland_altman(df, y_true, y_pred):
    mean_values = (df[y_pred] + df[y_true]) / 2
    diff_values = df[y_pred] - df[y_true]
    mean_diff = np.mean(diff_values)
    std_diff = np.std(diff_values, ddof=1)
    loa_upper = mean_diff + 1.96 * std_diff
    loa_lower = mean_diff - 1.96 * std_diff
    return mean_values, diff_values, mean_diff, std_diff, loa_upper, loa_lower
```
